```python
import re

def process_string(input_str):
    # Remove brackets and numbers using regex
    result_str = re.sub(r'\(?\s*\d*\s*\)?', '', input_str)
    
    # Remove extra whitespaces
    result_str = ' '.join(result_str.split())
    
    return result_str

# Example usage:
input_str = "three ( 3 )"
output_str = process_string(input_str)
print(output_str)

# Example usage:
input_str = "three ( 3 "
output_str = process_string(input_str)
print(output_str)
```

```python
import re

def get_words(text, words_to_remove_list):
    # Create a pattern to match any word that is in words_to_remove_list
    words_to_remove_pattern = r'\b(?:' + '|'.join(map(re.escape, words_to_remove_list)) + r')\b'
    
    # Extract only the words in words_to_remove_list
    matched_words = re.findall(words_to_remove_pattern, text, flags=re.IGNORECASE)
    
    if matched_words:
        cleaned_text = ' '.join(matched_words)
        cleaned_text = cleaned_text.strip()
        return cleaned_text
    else:
        return text

# Example usage:
text = "This is a sample text with some words to remove and some to keep."
words_to_remove_list = ['sample1', 'remove1']
result = get_words(text, words_to_remove_list)
print(result)
```

```python
import re
def preprocess_incoterm(text : str, incoterm_list):
    
	pattern = r'\b(?:' + '|'.join(re.escape(term) for term in incoterm_list) + r')\b'

	# Use re.findall to find all matching Incoterms in the text
	found_incoterms = re.findall(pattern, text, re.IGNORECASE)

	# If Incoterms are found, join them with spaces; otherwise, return the complete text
	result = " ".join(found_incoterms) if found_incoterms else text

	return result

preprocess_incoterm('CIP destination airport ( air freight ) CIP point of delivery ( truck )', list_of_incoterms)
```
